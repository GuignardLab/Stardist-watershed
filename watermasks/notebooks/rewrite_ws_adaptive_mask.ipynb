# Imports

```python
from lineagetree import LineageTree, read_from_mastodon
import numpy as np
import matplotlib.pyplot as plt
import h5py
from scipy.ndimage import gaussian_filter, median_filter, grey_erosion
from skimage.segmentation import watershed
from tqdm import tqdm

# CUSTOM FUNCTIONS
from watermasks.utils import paths, import_registration_mats_multiview, registered_position_of_id_in_t, paint_annotations, view_overlays_3d_V2, block_timer
from watermasks.main import percentile_intensities_sampler
```

```python
# --- IMPORT DATA AND FILES ---
path_h5, path_xml, path_mast, dir_input_images, dir_output_masks = paths('JR_22-10-18')
```

```python
VIEW='0'

# LOAD REG. MATRICES AND LINEAGETREE
R_of_t = import_registration_mats_multiview(path_xml, 0, 500)
lT = read_from_mastodon(path_mast)

TP = 300
# LOAD INPUT IMAGE
with h5py.File(path_h5, "r") as f5:
    test_image = f5[f't00{TP}']['s00']['2']['cells'][:]

shape_full_h5 = np.array([300, 1888, 2100]) # as measured in T7_3 data (dir = H:)
shape_test_image = np.array(np.shape(test_image))
scaling = shape_test_image/shape_full_h5
print(f'full image shape = {shape_full_h5}\n2nd h5 layer has shape = {shape_test_image} \n-scaling ratio is [z, y, x] = {scaling}')

# GET THE REGISTERED MASTODON POSITIONS
reg_pos_at_t = []
for mastodon_id_t in lT.time_nodes[TP]:
    x, y, z = registered_position_of_id_in_t(mastodon_id_t, lT, TP, R_of_t, VIEW, scaling)
    reg_pos_at_t.append([x, y, z])
reg_pos_at_t = np.asarray(reg_pos_at_t)

# --- VISUALIZE THE ANNOTATIONS ---
an_space = paint_annotations(TP, lT, shape_test_image,R_of_t,scaling,VIEW,1)
```

```python
# --- IMAGE PRE-PROCESSING ---

# PARAMETERS
EROSION_RADIUS = 7 # after tweaking 

im = test_image
pos_array = np.array([p[::-1] for p in reg_pos_at_t]).round().astype(np.uint16)
im_gauss = gaussian_filter(im, sigma=1)
im_gauss_median = median_filter(im_gauss, size=1) # this seems that is not doing much! (inspection by eye)

erosion = grey_erosion(im_gauss_median, size=EROSION_RADIUS)
im_for_ws = im_gauss_median - erosion
im_for_ws = gaussian_filter(im_for_ws, sigma = 1)
```

```python
#PARAMETERS
MIN_INT=10 # intensity is never accepted bellow this value as valid.
MAX_INT=500 # first intensity floor, to be lowered
MIN_VOL=300
MAX_VOL=1500
print_stats = True

from beautifultable import BeautifulTable

# USE ANNOTATION POSITION AS WS SEEDS
pos_array = np.array([p[::-1] for p in reg_pos_at_t]).round().astype(np.uint16)
seeds = np.zeros_like(im)
seeds[tuple(pos_array.T)] = np.arange(1, len(reg_pos_at_t) + 1) # each seed has it's own label
all_seeds = set(np.unique(seeds)) - {0} # don't include background

# MASK (TO BE UPDATED)
def get_mask(thresh_val):
    return (im_for_ws > thresh_val)

# INITIALIZE VARIABLES
ws = np.zeros_like(im)
vols = {}
missed_seeds = all_seeds

intensities = percentile_intensities_sampler(im_for_ws,intensity_thresh_range=[MIN_INT, MAX_INT], percentages=[90, 80]) # calculates the list of thresholds
intensities[100]=MAX_INT # for first iteration

# FOR STATS
if print_stats == True:
    for_table = []

# ADAPTIVE MASKS - MAIN LOOP
for pct, current_threshold in tqdm(sorted(intensities.items())[::-1],
        desc="Processing thresholds",
        unit="step"):
    
    if not missed_seeds: # break if previous step left no missing seeds
        break

    mask = get_mask(current_threshold)
    new_ws = watershed(np.max(im_for_ws) - im_for_ws, seeds, mask=mask)

    unique, counts = np.unique_counts(new_ws)
    vols = dict(zip(unique, counts))

    for s in missed_seeds.intersection(vols.keys()):
        if s != 0 and MIN_VOL < vols[s] < MAX_VOL: # update ws, only for acceptable volumes
            ws[new_ws == s] = s 

    if print_stats == True:
        n_found_seeds = len(missed_seeds) - len(set(np.unique(ws)))

    missed_seeds = all_seeds - set(np.unique(ws))
    
    if print_stats == True:
        n_missed_seeds = len(missed_seeds)
        for_table.append([f'{pct} %', current_threshold, n_found_seeds, n_missed_seeds])
```

```python
if print_stats == True:
    table = BeautifulTable()
    table.column_headers = ['Percentage', 'Percentile Value', 'Found Seeds', 'Missed Seeds']
    tt = np.asarray(for_table)
    for i in range(len(for_table)):
        table.rows.insert(i, for_table[i])
    print(table)
```
